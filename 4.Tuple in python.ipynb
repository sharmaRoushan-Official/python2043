## 4. Tuple in Python

### 1. What is a Tuple?

A `tuple` is a collection data type in python used to store multiple values in a single variable.

----

Key chracteristics:

* Ordered Collection
* Immutable(Cannot be changed after creation)
* Allow duplication items
* Can store different data types
* Faster than lists
* Written using round brackets()

```python
### Empty tuple
x = ()
print(x,type(x))
```

### 2. Creating Tuples

> There are only 3 method

> 1.()

> 2., (comma)

> 3.constructor-tuple()

Method-1

```python
fruits = ("Mango","Apple","Grapes","Strawberry")
print(fruits)
print(type(fruits))
```

Method - 2

```python
colors = "Black","Pink","Purple","Red"
print(colors)
print(type(colors))
```

Method-3  Using Constructor- tuple()- method

```python
numbers = tuple([1,2,3,4,5,6])
print(numbers)
print(type(numbers))
```

```python
numbers = tuple({1,2,3,4,5,6})
print(numbers)
print(type(numbers))
```

### 4. Nested Tuple: Tuple Inside another tuple



```python
nested_tuple = ((1,2),(3,4),(5,6),(7,8))
print(nested_tuple)
print(type(nested_tuple))
```

### Positive Indexing 

```python
nested_tuple = ((1,2),(3,4),(5,6),(7,8))
print(nested_tuple)
print(type(nested_tuple))

print(nested_tuple[0])
print(nested_tuple[1])
print(nested_tuple[2])
print(nested_tuple[3])
```

```python
# Negative Indexing
print(nested_tuple[-1])
print(nested_tuple[-2])
print(nested_tuple[-3])
print(nested_tuple[-4])
```

### 7. Tuple Immutability

Tuple cannot be modified or changed once it is created.

```python
numbers = (10,20,30,40,50,60,70,80,960,100)
print(numbers)
```

```python
numbers[0] = "Disha"
print(numbers)
```

```python
numbers = [10,20,30,40,50,60,70,80,960,100]
print(numbers)

numbers[0] = "Disha"
print(numbers)
```

### 8. Changing Tuple elements

```python
numbers = (10,20,30,40,50,60,70,80,960,100)  # immutable
print(numbers)
```

```python
temp = list(numbers)   # mutable
print(temp,type(temp))

temp[0] = "Pratiksha"
print(temp)
```

```python
num = tuple(temp)  # immutable: address of variable will be changed
print(num)
```

## 12. Tuple Uncakcing

```python
data = ("Python",25,True, False)
print(data)


a,b,c,d = data
print(a)
print(b)
print(c)
print(d)
```

```python
data = ("Python",25,True, False)
print(data)
```

### 13. Extended Unpacking 

Using *Operator (Asterisk)

```python
data = ("Python",25,True, False)
print(data)

a,b,c,d = data
print(a)
print(b)
print(c)
print(d)
```

```python
data = ("Python",25,True, False)
print(data)

a,*b,c = data
print(a)
print(b)
print(c)
```

```python
data = ("Python",25,True, False)
print(data)

*a,b,c = data
print(a)
print(b)
print(c)
```

```python
data = ("Python",25,True, False)
print(data)

a,b,*c = data
print(a)
print(b)
print(c)
```

### 17. Tuple Method

Tuples have only two built-in methods

1. count()



```python
numbers = (10,20,30,40,50,60,60,60,70,80,80)
print(numbers)
```

```python
numbers.count(80)
```

2. index()


```python
numbers = (10,20,30,40,50,60,60,60,70,80,80)
print(numbers)

numbers.index(80)
```

### Built-in Functions with tuples


```python
numbers = (10,20,30,40,50,60,60,60,70,80,80)
print(numbers)

print(max(numbers))
```

```python
print(min(numbers))
```

```python
numbers = (5,78,99,55,66,22,10,20,30,40,50,60,60,60,70,80,80)
print(numbers)
print(sorted(numbers))
```

### 20. Advantages of Tuples

1. Faster than List
