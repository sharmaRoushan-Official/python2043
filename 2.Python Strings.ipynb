### 2. Python String 

-----

Q. What is a string in Python?

A `String` in Python is a sequence of Unicode character enclosed in quotes.



```python
s1 = "Hello"
s2 = 'world'
s3 = """Multi-line String"""
print(s1)
print(s2)
print(s3)
```

### Key Points:

* String is Immutable(Can't changer after created)
* Index Value: Positive and Negative
* Supports slicing, iteration, and many built-in methods/functions



### 2. Creating String

Different ways

```python
s1 = "hello"  # double quotes
s2 = 'Hello'  # single quotes
s3 = '''Hello world'''   # triple quotes
print(s1)
print(s2)
print(s3)
```

### 3. String Indexing

Each character has a position/address

```python
s = "Python"
print(s)
```

|-|P|y|t|h|o|n|
|---|---|---|---|---|---|---|
|Positive|0|1|2|3|4|5|
|Negative|-6|-5|-4|-3|-2|-1|

```python
# positive indexing
s = "Python"
print(s[0])
print(s[1])
print(s[2])
print(s[3])
print(s[4])
print(s[5])
```

```python
### Negative Indexing

s = "Ptyhon"
print(s[-1])
print(s[-2])
print(s[-3])
print(s[-4])
print(s[-5])
print(s[-6])
```

```python
### Checking Methods:

## isalpha(): only alphabetical character 
## isdigit(): only digit character
## isalnum(): alphetical chracter and numerical value
## isspalce(): space value
## islower(): lower chracters
## isupper(): upper character
```

### Triming: White space remove
* strip() : both side trim white space
* lstrip(): left side trim white space
* rstrip(): right side trim white space



```python
s = "            Hello world                  "
print("Original String: ",s)

s1 = s.strip()
print(s1)

s2 = s.lstrip()
print(s2)

s3 = s.rstrip()
print(s3)
```

### Replace()  <- remove old value and add new value on same position


```python
s = "Hello world"
print("Original String:",s)

s1 = s.replace("world","Pratiksha")
print(s1)


s2 = s.replace("Hello","Hey")
print(s2)
```
