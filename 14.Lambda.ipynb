### 14. Lambda functions in Python

##### What is Lambda functions?
A Lmabda function is small, anonymous function[without their name or nameless] written in one single line.


📌 Simple Definition

> A lambda function is one-line function without a name.

### Why Lambda function are used?

* ꪜShort & Clean Code
* ꪜUsed for temporary operations
* ꪜCommon with map(), filter(), reduce(), apply(),sort()
* ꪜImproves readablity in simple logic

### Normal Functions VS Lambda functions


#### Normal Functions


```python
## user-defined function
def square(x):
    return x*x

square(10)
```

```python
square(100)
```

##### Lambda Functions: Nameless

```python
(lambda x:x*x)(10)
```

```python
(lambda y:y*y)(100)
```

Notes:

* ꪜSame Output
* ꪜLess Code

#### Syntax of Lambda functions
`(lambda argument:expression)`

##### 📌 Rules:

* ꪜCan have multiple arguments
* ꪜOnly one expression
* ꪜAutmatically returned value
* ꪜNo return keyword

#### Basic Examples:


1. Add Two Numbers:



```python
# Normal Functions 

def sumTwo(a,b):
    sumValue = a+b
    return sumValue

sumTwo(10,20)
```

```python
### anonymous function or lambda function
(lambda a,b : a+b)(10,20)
```

### 2. Square of a Number



```python
# Normal Function 

def sqrNum(num):
    return num**2
sqrNum(10)
```

```python
# lambda function 

(lambda a:a**2)(10)
```

### 3. Check Even or Odd(Condition)

```python
def checkEven(a):
    if a%2==0:
        return f"Even Number: {a}"
checkEven(10)
```

```python
### Lambda function

(lambda x:x%2==0 and "Even")(10)
```

```python
### Lambda function

(lambda x:x%2==0 and "Even")(11)
```

```python
def checkEven(a):
    if a%2==0:
        return f"Even Number: {a}"
    else:
        return f"Odd Number: {a}"
checkEven(11)
```

```python
(lambda x:"Even" if x%2==0 else "Odd")(10)
```

```python
(lambda x:"Even" if x%2==0 else "Odd")(11)
```

### Lambda with Multiple Arguments

```python
def maxThree(a,b,c):
    if a>b and a>c:
        print(a, "is greater")
    elif b>a and b>c:
        print(b, "is greate")
    else:
        print(c,"is greater")
maxThree(11,10,12)
```

Lambda with Filter()

prupose:
Filter elements based on conditons

```python
numbers = [10,11,12,15,14,16,18,17,19,20,22,24,26,23,25,2,72,82,88,89]


def evenNum():
    for i in numbers:
        if i%2==0:
            print(i,"Even")
        else:
            print(i,"Odd")
evenNum()
```

### Lambda functions

syntax:
filter(func, sequence)

```python
num = [10,11,12,15,14]

def evenNum(x):
    if x%2==0:
        return "Even"
```

```python
list(filter(evenNum,num))
```

```python
tuple(filter(evenNum,num))
```

### lambda with filter

```python
list(filter(lambda a:a%2==0,num))
```

#### Lambda with Map

Purpose:

Apply a function to each element in the list.


syntax:

`map(function,sequence)`

```python
num = [10,11,12,15,14]

def sqrNum():
    for i in num:
        print(i,":",i*i)
sqrNum()
```

```python
def sqrNum(x):
    return x*x
sqrNum(10)
```

```python
num = [10,11,12,15,14]
list(map(sqrNum,num))
```

```python
### with Lambda function
num = [10,11,12,15,14]
list(map(lambda i: i * i,num))
```

### lambda with reduce()

📌 required functools library

syntax:

reduce(function, sequence)

```python
from functools import reduce
```

```python
num = [10,11,12,15,14]  # add, sub, multiple, divide

reduce(lambda a,b:a+b,num)
```

```python
num = [10,11,12,15,14]  # add, sub, multiple, divide

reduce(lambda a,b:a-b,num)
```

```python
def redNum(a,b):
    return a+b
```

```python
num = [10,11,12,15,14] 
reduce(redNum,num)
```

```python
#lambda
num = [10,11,12,15,14] 
reduce(lambda a,b:a+b,num)
```

### Lambda with Sort()

syntax:

`sorted(sequence, function)`

```python
students = [("Ramesh",66,65),("Anita",92,17),("Ankita",55,108)]
print(students)
```

```python
sorted(students, key = lambda x:x[2])
```
