## 6. Dictionary in Python

Q. What is Dictionary in Python?

---
A dictionary is a collection of key:value pairs.

* Each key is unquie.
* Each key maps to a value.
* Dictionary are mutable(changeable)
* It is Written in {}

```python
d1 = {}
print(d1,type(d1))
```

```python
x1 = set()
print(x1)
print(type(x1))
```

```python
student = {"name":"Arnav",'age':25,"course":"Python"}
print(student)
```

```python
student = {
    "name":"Arnav",
    'age':25,
    "course":"Python"
}
print(student)
```

### 11. Looping Through dictionary

```python
dict1 = {'name': 'Disha', 'age': 20, 100: 'Number', 'Address': 'Pune'}
print(dict1)
```

```python
for key in dict1:
    print(key)
```

```python
for value in dict1.values():
    print(value)
```

```python
for key,value in dict1.items():
    print(key,":",value)
```
