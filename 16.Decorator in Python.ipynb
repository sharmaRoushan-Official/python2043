Q. What is Decorator?

A `decorator` is a funtion that modifies antoher function's behavior without changing its code.

📌 Define
> A decorator adds extra feature to a function.

### Why Decorator are Needed

❌Withou Decorator

if you want to:
* Log functions calls
* Check user authentication
* Measure execution time

You would have to repeate ths same code in every function.

✅ With Decorator

* Write the logic once
* Resuse it many times
* Cleaner and maintainble code.


```python
def decorator(func): # x
    def wrapper():
        print("Before Functions")
        func()
        print("After Function")
    return wrapper
```

```python
def say_hello():
    print("Hello world")
    # return "Hello world"
say_hello()
```

### Applying a Decorator(Manual Way)

```python
val = decorator(say_hello)
val()
```

### Using @ Syntax 

```python
@decorator
def greet():
    user = input("Enter your name:")
    greet = "Congratulations!! "+user
    print(greet)
greet()
```

### Decorator with Arguments


```python
def decorator(func):
    def wrapper(name):
        print("Congratulations!!")
        func(name)
    return wrapper
```

```python
@decorator
def greet(name):
    print("Hello ",name)
greet("Akash")
```

#### Decorator with *args and **kwargs

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function Started")
        result = func(*args, **kwargs)
        print("Function ended")
    return wrapper
```

```python
import time
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function Started")
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Execution Time",end-start)
        print("Function ended")
    return wrapper
```

```python
@decorator
def demo():
    for i in range(1000000000):
        pass
demo()
```

### Decorator with Return Value



```python
def decorator(func):
    def wrapper():
        return func() + "!!!!"
    return wrapper
```

```python
@decorator
def message():
    return "Hello"
message()
```

#### Multiple Decorator

```python
def decor1(func):
    def wrapper():
        print("Decor 1")
        print(func())
    return wrapper


def decor2(func):
    def wrapper():
        print("Decor 2")
        func()
    return wrapper
```

```python
@decor2
@decor1
def show():
    return "Hello world"
show()
```

```python
# 4. Function Call Counter
# Create a decorator that counts and displays how many times a function has been called.

def call_counter(func):
    count = 0

    def wrapper():
        nonlocal count
        count += 1
        print("Function called:", count, "times")
        func()

    return wrapper


@call_counter
def greet():
    print("Hello, welcome to Python!")
```
