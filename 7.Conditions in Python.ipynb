### 7. Conditions in python

1. What is a conitional Statemenet?

A `conditional statemenent` allows a program to makr decision based conditions(True/False)
* if
* elif
* else

### 2. Basic Syntax of if condition

syntax: 

```text
if condition:
    # code execute if condition is True

````python
age = 16

if age>=18:
    print("you can vote")
````

Note:

* Condition must return True/False
* Indentation is mandatory
* No parentheses required(unlike c/java)

````python
age = 20

if (age>=18):
    print("you can vote")
````

### 3. if-else statement

````python
age = 16

if age>=18:
    print("you can vote")  # True Block
else:
    print("You cannot vote") # False Block
````

Note:

* else runs when condition is False
* only one else allowed per if

### 10. Truth and Falsy Values

-------------

False Values:

* False
* None
* 0
* "" or ''
* [],(),{}


````python
name = ""
if name:
    print("Name Exists")
else:
    print("Name not exists")
````

````python
name = False
if name:
    print("Name Exists")
else:
    print("Name not exists")
````

````python
name = 0
if name:
    print("Name Exists")
else:
    print("Name not exists")
````

````python
name = []
if name:
    print("Name Exists")
else:
    print("Name not exists")
````

````python
name = ()
if name:
    print("Name Exists")
else:
    print("Name not exists")
````

````python
name = {}
if name:
    print("Name Exists")
else:
    print("Name not exists")
````

## 13. Advances Case



````python
day = int(input("Enter any number from 1 to 7: "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Weekday")
````
