### 17. Generator in Python

#### Q. What is a Generator?
A generator is a special function that returns an iterator and provides values one at a time using yield instead of return.

📌Define:
> A generator generates values one by one, only when needed.

#### Why Generator are needed?

❌Without Generator


```python
def get_numbers():
    return [1,2,3,4,5,6]
get_numbers()
```

Notes:

All values stored in mememory at once.

✅️ with Generator

```python
def get_numbers():
    for i in range(1,6):
        yield i
x = get_numbers()
print(x)
```

```python
print(next(x))
```

```python
print(next(x))
```

```python
print(next(x))
```

```python
print(next(x))
```

```python
print(next(x))
```

```python
print(next(x))
```

Notes:

* ꪜValue created on demand
* ꪜMemory efficient

#### Generator with Loop 

```python
def count_up(n):
    for i in range(1,n+1):
        yield i
```

```python
for num in count_up(6):
    print(num)
```

```python
obj = count_up(6)
print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

### How Generator Works


📌 Steps:

1. Function is called ➼ No Execution
2. Return generator obj
3. next() resumes execution
4. Stop at yield
5. contunues from same point


#### Generator VS Normal Function

----

|Feature|Generator|Function|
|-----|-----|------|
|Used|`yield`|`return`|
|Memory|`Low`|`High`|
|Execution|`Lazy`|`Immediate`|
|Values|`One at a time`|`All at once`|

### Example: Even Numbers using Generator



```python
def evenNum(n):
    for i in range(n):
        if i%2==0:
            yield i
```

```python
for i in evenNum(10):
    print(i)
```

```python
obj = evenNum(10)
print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

### Reading Files Uisng Generator



```python
def readfile(fileName):
    with open(fileName,'r') as file:
        for line in file:
            yield line.strip()
```

```python
data = readfile(r"C:\Users\User\Desktop\NursaryRhymes.txt")
print(next(data))
```

```python
print(next(data))
```

```python
print(next(data))
```

```python
print(next(data))
```

```python
print(next(data))
```

### Generator with send()



```python
def gen():
    value = yield 1  # first assign 1
    yield value  # value None
g = gen()
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
g.send(10)
```
