## ATM Project

class Atm:
    def __init__(self,accountNumber,balance=0):
        while True:
            self.name = input("Set Your Name: ")
            if self.name.isalpha():
                print("Name Added Successfully",self.name)
                break
            else:
                print("Invalid Name! Enter a valid Name.")
            
                
        self.accountNumber=accountNumber ## 10 digit only
        try:
            while True:
                self.pin = input("Set your PIN:")
                if self.pin.isnumeric and len(self.pin) == 4:
                    print("PIN Added SuccessFully")
                    break
                else:
                    print("Incorrect PIN")
        except Exception as e:
            print("Error:",e)
        
        self.balance = balance

    def accountDetails(self):
        """Display Account Details"""
        print("================================")
        accountNumberstart = self.accountNumber[:3] #0,1,2
        accountNumberEnd = self.accountNumber[-2:]
    
        print(f"Customer Name is:{self.name}")
        print(f"Customer Account Number: {accountNumberstart+"XXXXXX"+accountNumberEnd}")
        print("Your Current PIN:'XXXX'")
        print(f"Your Current Balance: {self.balance}")
        print("================================")

    def checkBalance(self):
        pin2 = input("Enter your current pin: ")
        if self.pin == pin2:
            print(f"Your Current Balance: {self.balance}")
        else:
            print('Invalid PIN! Enter Correct PIN')


    def deposite(self):
        pin2 = input("Enter your Current PIN: ")
        if self.pin == pin2:
            amount = int(input("Enter your amount ₹: "))
            if amount>0:
                self.balance = self.balance+amount
                print(f"Your Current Balance Updated ₹: {self.balance}")
            else:
                print("Invalid Amount Enter a Valid Amount")
        else:
            print("Invalid PIN! Transection Failed.")

    def withdraw(self):
        pin2 = input("Enter PIN: ")
        if self.pin == pin2:
            amount = int(input("Enter Amount: "))
            if amount<self.balance: 
                self.balance = self.balance - amount
                print("Collect Your Cash:",amount)
                print("Current Balance is:",self.balance)
            else:
                print(f"Insufficient Funds!!\nCurrent Amount:{self.balance}")
        else:
            print("Invalid PIN")
            
    def changeName(self):
        """Change Customer Name"""

        pin2 = input("Enter Your Current PIN: ")
        if self.pin == pin2:
            newName = input("Enter Your New Name: ")
            if newName.strip() != "":
                self.name = newName
                print("Name Changed Successfully.")
                print(f"New Customer Name: {self.name}")
            else:
                print("Name Cannot Be Empty.")
        else:
            print("Invalid PIN! Name Change Failed.")


    def trans(self):
        print("""
        \n1.Account Details
        \n2.Account Balance
        \n3.Deposite Cash
        \n4.Withdraw Amount
        \n5.Change Name
        \n6.Show Account Number
        \n7.Change PIN
        \n8. Exit""")
        while True:
            try:
                opt = int(input("\nChoose Any Option: "))
            except Exception as e:
                print("Error:",e)
                continue
            else:
                if opt == 1:
                    Atm.accountDetails(self)
                elif opt == 2:
                    Atm.checkBalance(self)
                elif opt == 3:
                    Atm.deposite(self)
                elif opt == 4:
                    Atm.withdraw(self)
                elif opt == 5:
                    Atm.changeName(self)
                elif opt == 8:
                    print("Thanks for using ATM")
                    break
