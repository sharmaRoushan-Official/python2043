### 3. Lists in Python

* A `List` in Python is a mutable(Can be changes), Ordered, Indexed Collection that can store multiple values.
* Inclduing (mixed data types)
* Lists are one of the most used data types in Python

1. Syntax

`myList = [1,2,3,4,5]`

```python
my_list = [1,2,3,4]
print(my_list)
```

```python
data = [10,20,30,40,50]
print(data,type(data))
```

### 2. Features of List

----

|Feature|Explanation|
|-----|----|
|`Ordered`| Items maintain insertion order|
|`Indexed`| Each element has an index position|
|`Mutable`| Can be modified after creating|
|`Allow Duplicate`| Same values can repeat|
|`Mixed type allowed`| int, float, str, bool, object-[list, tuple,set, dictionary]|

### Creating List

```python
x = [10,52.36,True,False, "Python",[10,20],(10,15,14),{10,45},{"key_1":45}]
print(x)
```

## 3. List Indexing

1. Positive Indexing
2. Negative Indexing

```python
x = [10,52.36,True,False, "Python",[10,20],(10,15,14),{10,45},{"key_1":45}]
print(x)

print(x[0])
print(x[1])
print(x[2])
print(x[3])
print(x[4])
print(x[5])
print(x[6])
print(x[7])
print(x[8])
```

```python
## Negative Indexing

x = [10,52.36,True,False, "Python",[10,20],(10,15,14),{10,45},{"key_1":45}]
print(x)
print(x[-1])
print(x[-2])
print(x[-3])
print(x[-4])
print(x[-5])
print(x[-6])
print(x[-7])
print(x[-8])
print(x[-9])
```

### 4.Slicing of List

### 10. sort()


Ascending or descending order

```python
nums = [10,20,30,30,40,50,60,50,60,60,60,-70,10,5,8,90.36,15]
print(nums)
```

```python
nums.sort()  # ascending order
print(nums)
```

```python
## descending order

nums.reverse()
print(nums)
```

```python
nums = [10,20,30,30,40,50,60,50,60,60,60,-70,10,5,8,90.36,15]
print(nums)
```

### Copy() 

1. Deep Copy [=] --> Address Value same

2. Shallow copy [.copy()] --> Different address Value

```python
x = [10,20,30]
print("Original Value:",x)
```

```python
id(x)
```

```python
# Deep Copy [=] --> Address Value same

y = x
print(y)
```

```python
id(y)
```

```python
x.append("Disha")
print(x)
```

```python
print(y)
```

## shallow copy() 

```python
x = [10,20,30]
print("Original Value:",x)
```

```python
y = x.copy()
print(y)
```

```python
id(x)
```

```python
id(y)
```

```python
x.append("Disha")
print(x)
```

```python
print(y)
```
