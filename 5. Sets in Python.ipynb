## 5. Sets in Python
------
1. Introduction of Sets

A `Set` in Python is a collection of data type used to store multiple unique element.

It is based on mathematical concept of set.

Key Feature:

1. Unordered
2. Unindexed
3. Mutable(Can be change)
4. Does not allowed duplicate value
5. Elements must be immutable(hashable)  

```python
# empty dictionary
s = {} # dictionary
print(s,type(s))
```

```python
# empty dictionary
s = {1} # dictionary
print(s,type(s))
```

```python
s1 = set()
print(s1,type(s1))
```
