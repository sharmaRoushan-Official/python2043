## 5. Sets in Python
------
1. Introduction of Sets

A `Set` in Python is a collection of data type used to store multiple unique element.

It is based on mathematical concept of set.

Key Feature:

1. Unordered
2. Unindexed
3. Mutable(Can be change)
4. Does not allowed duplicate value
5. Elements must be immutable(hashable)  

```python
# empty dictionary
s = {} # dictionary
print(s,type(s))
```

```python
# empty dictionary
s = {1} # dictionary
print(s,type(s))
```

```python
s1 = set()
print(s1,type(s1))
```

## 10. Set operations

##### Python set supports mathematical operations

1. union
2. intersection
3. difference
4. symmetric_difference

```python
A = {1,2,3}
B = {3,4,5}

# {1,2,3,4,5}
# mthod 
# symbol 
print(A)
print(B)
```

```python
# mthod 
# symbol

print(A.union(B))
```

```python
# symbol 

print(A|B)
```

2. Intersection

```python
A = {1,2,3}
B = {3,4,5}

# {1,2,3,4,5}
# mthod 
# symbol 
print(A)
print(B)
```

```python
print(A.intersection(B)) # method
```

```python
print(A & B)  # symbol
```

```python
A = {1,2,3}
B = {3,4,5}

# {1,2,3,4,5}
# mthod 
# symbol 
print(A)
print(B)

```

```python
## Difference

# A
# B

# A-B
# B-A
```

```python
A = {1,2,3}
B = {3,4,5}

# {1,2,3,4,5}
# mthod 
# symbol 
# print(A)
# print(B)

print(A.difference(B))
print(B.difference(A))
```

```python
print(A-B)
```

```python
print(B-A)
```

### Symmetric Difference

```python
A = {1,2,3}
B = {3,4,5}

# {1,2,3,4,5}
# mthod 
# symbol 
print(A)
print(B)
```

```python
# method

print(A.symmetric_difference(B))
```

```python
# symbol 

print(A^B)
```

## 18. When to use Sets.

Use sets when:

* ✅ Remove duplicate
* ✅ Fast Searching
* ✅ Mathematical Operations: Union, Intersection, difference, symmetric_difference, disjoint,frozenset,subset,superset
* ✅ Membership testing

Example: 
* Unique email id
* Unique customer id
* Tag System
* Data Filtering
