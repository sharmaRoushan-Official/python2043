#### Convert Python ---> JSON

```python
data = {
    "name":"Pratiksha",
    "age":85,
    "course":"Python Programming"
}
print(data)
```

Convert Python Dictionary  --->: JSON

```python
json_data = json.dumps(data)
print(json_data)
```

```python
json_data[0]
```

#### 2. Convert json -> Python Object

```python
json_data = '{"name": "Pratiksha", "age": 85, "course": "Python Programming"}'
print(json_data)

pythonDict = json.loads(json_data)
print(pythonDict)
```

1. dumps() ---- Python dictionary -> JSON
2. loads() ----> JSOn  ----> Python Dictionary

#### 3. Working with JSON files

### Write JSON file

```python
data = {
    "name":"Pratiksha",
    "age":58
}
print(data,type(data))
```

```python
with open("data.json",'w') as file:
    json.dump(data,file,indent=4)
```

### dump() vs dumps() and load() vs loads()

-----

|Function|Use()|
|---|----|
|`dump()`| Write File|
|`dumps()`|Convert Python dictionary to JSON string|
|`load()`| Read file|
|`loads()`| Convert string to dictionary|

```python
response = '{"status":"success","code":200}'
print(response,type(response))
```

```python
data = json.loads(response)
print(data,type(data))

if data["status"] == "success":
    print("API is working")
else:
    print("API is not working")
```

### Handlig JSON Error

```python
try:
    # data = json.loads(response)
    data = json.loads("Invalid Code")
except json.JSONDecodeError:
    print("Invalid JSON Format")
else:
    print(data)
finally:
    print("Code Running Successfully")
```

```python
try:
    # data = json.loads(response)
    data = json.loads("Invalid Code")
# except json.JSONDecodeError:
#     print("Invalid JSON Format")
except Exception as e:
    print("Error:",e)
else:
    print(data)
finally:
    print("Code Running Successfully")
```
