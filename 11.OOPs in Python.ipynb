### a). Single/Simple Inheritance

Meaning:
* In `Single Inheritance`, a child inherits from one parent class.

📌 Structure
`parent ➜ child`

### Why it is used:
* Code resuability
* Child class can use parent class properties and methods.
* Easy ot understand and maintain

### Real-Life Example:

* Parent -> Vehicle
* Child -> Car

A Car is a Vehicle

Python Example

````python
class Vehicle:  # parent
    def start(self):
        print("Vehicle Started")

        
class Car(Vehicle):  # drived class/ child class
    def drive(self):
        print("Car is driving")


# obj = Vehicle()
# obj.start()

obj1 = Car()
obj1.drive()  # drive is the method of child class
obj1.start() # start is the method of parent class
````

Note:

Single Inheritance is most basic and safest inheritance. It avoids complexity and ambiguity.

### b). Multiple Inheritance

Meaning:

In `Multiple Inheritance`, a child class inherits from more than one parent class.


📌 Structure

```
    Parent1          Parent2
         \            /
          \          /
           \        /
             child

```

#### Why it is used?

* When once class needs feature from multiple parent classes.

* Helps to combine functionalities.


#### Real-Life Example

* parent1:Camera
* parent2: Phone
* child: SmartPhone


SmartPhone is both a Phone and a Camera

````python
class Phone:
    def call(self):
        print("Calling.......")

class Camera:
    def click_photo(self):
        print("Photo Clicked")


class SmartPhone(Camera,Phone):
    def internet(self):
        print("Google Connected....")

p1 = SmartPhone()
p1.call()
p1.click_photo()
p1.internet()
````

````python
print(SmartPhone.mro())
````

#### 3. Multilevel Inheritance 

Meaning:

In `Multilevel Inheritance`, inherits happens in levels.

📌 Stucture


```
GrandParent ➜ Parent ➜ Child
```

### Why it is used?
* To represents hierarchy in real world
* Step-by-step inheritance of feature

#### Real-Life Example

* Grandparent: Animal
* Parent: mammal
* child: Dog



````python
class Animal:  # parent
    def eat(self):
        print("Eating...")

class Mammal(Animal):
    def walk(self):
        print("Walking....")

class Dog(Mammal):
    def barks(self):
        print("Barking....")

d = Dog()
d.eat()
d.walk()
d.barks()
````

````python
print(Dog.mro())
````

#### 4. Hierarchical inheritance

Meaning:

In `Hierarchical Inheritance`, Multiple child inherits from a single parent class.


📌Structure

```
                        parent1
                       /   |   \
                child1  child2  child3
```

### Why it is used?

* One parent has common functionality
* Differnt children use and extend it different.

#### Real-Life Example

* Parent: Employee
* Children: 
    * Manager
    * Developer
    * Tester
    * HR

All are employees but have different roles.


Python Example

````python
class Employee:
    def work(self):
        print("All employees are working.....")


class Manager(Employee):
    def manage(self):
        print("Manager is managing work.....")

class Developer(Employee):
    def code(self):
        print("Developer is coding.....")

class Tester(Employee):
    def test(self):
        print("Tester is testing.......")

m = Manager()
m.work()
m.manage()

d = Developer()
d.work()
d.code()

t = Tester()
t.work()
t.test()
````
