#### 5. Constructor(`__init___`)


A constructor initializes object data automatically when the object is created.

````python
# int x;
# x = 20
````

````python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
````

````python
obj1 = Student("Disha",5)
obj1.name
````

````python
obj1.age
````

KeyPoint:

* Automatically Invoked
* Used to initialize instance variables

````python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def studentDetails(self):
        print(f"Myself {self.name}")
        print(f"I am {self.age} years old.")

obj2 = Student("Disha",85)
obj2.studentDetails()
````

````python
obj3 = Student("Pratiksha",9)
obj3.studentDetails()
````
