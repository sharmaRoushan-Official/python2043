### a). Single/Simple Inheritance

Meaning:
* In `Single Inheritance`, a child inherits from one parent class.

📌 Structure
`parent ➜ child`

### Why it is used:
* Code resuability
* Child class can use parent class properties and methods.
* Easy ot understand and maintain

### Real-Life Example:

* Parent -> Vehicle
* Child -> Car

A Car is a Vehicle

Python Example

````python
class Vehicle:  # parent
    def start(self):
        print("Vehicle Started")

        
class Car(Vehicle):  # drived class/ child class
    def drive(self):
        print("Car is driving")


# obj = Vehicle()
# obj.start()

obj1 = Car()
obj1.drive()  # drive is the method of child class
obj1.start() # start is the method of parent class
````

Note:

Single Inheritance is most basic and safest inheritance. It avoids complexity and ambiguity.

### b). Multiple Inheritance

Meaning:

In `Multiple Inheritance`, a child class inherits from more than one parent class.


📌 Structure

```
    Parent1          Parent2
         \            /
          \          /
           \        /
             child

```

#### Why it is used?

* When once class needs feature from multiple parent classes.

* Helps to combine functionalities.


#### Real-Life Example

* parent1:Camera
* parent2: Phone
* child: SmartPhone


SmartPhone is both a Phone and a Camera

````python
class Phone:
    def call(self):
        print("Calling.......")

class Camera:
    def click_photo(self):
        print("Photo Clicked")


class SmartPhone(Camera,Phone):
    def internet(self):
        print("Google Connected....")

p1 = SmartPhone()
p1.call()
p1.click_photo()
p1.internet()
````

````python
print(SmartPhone.mro())
````

#### 3. Multilevel Inheritance 

Meaning:

In `Multilevel Inheritance`, inherits happens in levels.

📌 Stucture


```
GrandParent ➜ Parent ➜ Child
```

### Why it is used?
* To represents hierarchy in real world
* Step-by-step inheritance of feature

#### Real-Life Example

* Grandparent: Animal
* Parent: mammal
* child: Dog



````python
class Animal:  # parent
    def eat(self):
        print("Eating...")

class Mammal(Animal):
    def walk(self):
        print("Walking....")

class Dog(Mammal):
    def barks(self):
        print("Barking....")

d = Dog()
d.eat()
d.walk()
d.barks()
````

````python
print(Dog.mro())
````

#### 4. Hierarchical inheritance

Meaning:

In `Hierarchical Inheritance`, Multiple child inherits from a single parent class.


📌Structure

```
                        parent1
                       /   |   \
                child1  child2  child3
```

### Why it is used?

* One parent has common functionality
* Differnt children use and extend it different.

#### Real-Life Example

* Parent: Employee
* Children: 
    * Manager
    * Developer
    * Tester
    * HR

All are employees but have different roles.


Python Example

````python
class Employee:
    def work(self):
        print("All employees are working.....")


class Manager(Employee):
    def manage(self):
        print("Manager is managing work.....")

class Developer(Employee):
    def code(self):
        print("Developer is coding.....")

class Tester(Employee):
    def test(self):
        print("Tester is testing.......")

m = Manager()
m.work()
m.manage()

d = Developer()
d.work()
d.code()

t = Tester()
t.work()
t.test()
````

````python
print(Manager.mro())
````

````python
print(Developer.mro())
````

````python
print(Tester.mro())
````

Note:

This is very common in software systems:

(Banking, College system, Employee Management)



### 5. Hybrid Inheritance


Meaning:
Hybrid Inheritance is combination of two or more inhertiance types:

It May Include
* a). Multiple + Mulilevel

* b). Hierarchical + Multiple



📌 Structure


```
           A
         /   \
        B     C
         \   /
           D
```

Here,
* B and C inherts from A (Hierarchical Inheritance)
* D inherits from B and C (Multiple)

### Why it is used?


* When a system requires complex relationship
* Common in large framework and enterprises projects.



Python Example

````python
class A:
    def showA(self):
        print("Class A")

class B(A):
    def showB(self):
        print("Class B")

class C(A):
    def showC(self):
        print("class C")

class D(B,C):
    def showD(self):
        print("class D")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
# obj.showE()
````

````python
print(D.mro())
````

Note:

Hybrid Inheritance is powerful but can cause confusin due to:
* Method conflict
* ambiguity

That's by python uses mro(Method Resolution Order)

### 2. Polymorphism

Meaning of Polymorphism

In OOPs the same method name behave differently depending on the object that calls it.


📌 Simple Definition
> _One Function or method, but multiple behaviors._


### Why Polymorphism is Important.

* Code become flexible
* Code become reusable
* Supports real-world scenarios
* Make program easy to extend
* Reduce code duplication

### Real-Life Example


Example: Person Roles

    * CEO
    * Manager
    * Employee
    * Developer
    * Team Leader
    * Tester
    * HR

### Note:

* Same person, different behavior based on role this is called polymorphism.


Types of Polymorphism:

1. Method Overloading(CompileTime Polymorphism)
   * Same method name different parameters
2. Method Overriding(Runtime Polymorphism)
   * Same Method name and Same Parameters
3. Polymorphism using Inheritance
4. Polymorphism using duck typing
5. Operator Overloding




### 1. Method Overloading(CompileTime Polymorphism)
Same method name different parameters


Note:

Python does not support traditional overloading, but we can achieve it using default arguments.


````python
class Calculator:
    def add(self,a,b=0,c=0):
        return a+b+c
obj = Calculator()
print(obj.add(10))  # run 
print(obj.add(10,20))  # add para - 2
print(obj.add(10,20,30))  # add para - 3
````

### KeyPints

* Same method name `add`
* Differet number of aruguments 
* Behaviour changes automatically



#### 2. Method Overriding (Same Signature)

Child class provides its own implemenetation of parent method



````python
class Employee:
    def role(self):
        return "I am an Employee"

class Trainer(Employee):
    def role(self):
        return "I am a Trainer"


class TeamLead(Trainer):
    def role(self):
        return "I am Leading the Team"

obj = TeamLead()
obj.role()
````

#### 3. Polymorphism using Inheritance

````python
class Shape:
    def area(self):
        pass

class Rectangle(Shape):
    def area(self):
        return "area = length* width"

class Circle(Shape):
    def area(self):
        return "area = π * r²"
````

````python
obj = Circle()
obj.area()
````

````python
shapes = [Rectangle(), Circle()]
for shape in shapes:
    print(shape.area())
````

### 4. Polymorphism using duck typing

📌 Rule:

> If it looks lick a duck and quacks like a duck, it is a duck.


````python
class Trainer:
    def work(self):
        print("Trainer teaches students")

class Developer:
    def work(self):
        print("Developer writes code.")

class Tester:
    def work(self):
        print("Tester testing code")

def do_work(emp):
    emp.work()
````

````python
do_work(Trainer())
````

````python
do_work(Developer())
````

````python
do_work(Tester())
````
