### Mode and meaning


|`Mode`|`Meaning`|
|---|---|
|`r`|Read(default)|
|`w`|Write|
|`a`|append|
|`r+`|Read and write|

Method 1 :read() ---> read entire file at once

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(path)

# Open the file

file = open(path,'r')


# Read the file
content = file.read()
print(content)

# close the file
file.close()
````

### Meathod 2: Readline() ---> Read data one line at a time.

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(path)

# Open the file

file = open(path,'r')


# Read the file
content = file.readline()
content1 = file.readline()
content2 = file.readline()

print(content)
print(content1)
print(content2)
# close the file
file.close()
````

### Method 3: readlines() --> Read all lines as list()

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(path)

# Open the file

file = open(path,'r')


# Read the file
content = file.readlines()
print(content)

# close the file
file.close()
````

### Reading Files using Loop



````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(path, 'r')

for line in file:
    print(line)
file.close()
````

## Using With Statement [Recommendation]


* file automatically closed by system

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"


with open(path,'r') as f:
    print(f.read())
````

````python
with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as obj:
    print(obj.read())
````

````python
with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as f:
    # print(f"{f.read()}")
    char=0
    space = 0
    data1  = f.read()
    for data in data1:
        if data == " ":
            space+=1
        char+=1
        # print(data)
    print("Total Characters is:",char)
    print("Total Spaces:",space)
    print("total chracters without spaces:",char-space)
````
