Method 1 :read() ---> read entire file at once

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(path)

# Open the file

file = open(path,'r')


# Read the file
content = file.read()
print(content)

# close the file
file.close()
````

### Meathod 2: Readline() ---> Read data one line at a time.

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(path)

# Open the file

file = open(path,'r')


# Read the file
content = file.readline()
content1 = file.readline()
content2 = file.readline()

print(content)
print(content1)
print(content2)
# close the file
file.close()
````

### Method 3: readlines() --> Read all lines as list()

````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
print(path)

# Open the file

file = open(path,'r')


# Read the file
content = file.readlines()
print(content)

# close the file
file.close()
````

### Reading Files using Loop



````python
path = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

file = open(path, 'r')

for line in file:
    print(line)
file.close()
````
