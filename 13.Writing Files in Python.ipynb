### 13. Writing Files in Python


Q. What is File Writing?

`Writing Files` means storing data permenently in a file from a Python programming.



📌 Used For?

* Saving user's data
* Logs
* Exporting result
* Backup files


----

### File Writing Modes

| Mode| Meaning|
|----|-----|
|`w`|Write(Create/Overwrite)|
|`a`|append (adds data at end)|
|`w+`| Write+Read|
|`a+`|Append and read|
|`x`|Creates if file not exists and throw error if file already exists|

----

### Step by step process to write file

1. Open the file
2. write data in the file
3. close the file

```python
## write() mode
fileName = r"C:\Users\User\Downloads\disha.txt"

# step-1
file = open(fileName,'w') 


# step - 2

file.write("Myself Disha please help me!(bacchav)\n")
file.write("I am 30 years old")

# step-3
file.close()
```

```python
## write() mode
fileName = r"C:\Users\User\Downloads\disha.txt"

# step-1
file = open(fileName,'a') 


# step - 2

file.write("\nMyself Disha please help me!(bacchav)\n")
file.write("I am 30 years old")

# step-3
file.close()
```

```python
fileName = r"C:\Users\User\Downloads\disha1.txt"


with open(fileName,'x') as f:
    f.write("Myself Please Help me.\n")
    f.write("I am in the trouble.")
```

### 5. Write + Read (Ww+)



```python
with open("sample.txt","w+") as file:
    file.write("Hello world")
    file.seek(0)
    print(file.read())
```

### 6. Writing using Loop(Log File)



```python
with open("log.txt",'w') as file:
    for i in range(3):
        file.write(f"Log entry {i}\n")
```

## 7. Exception handling while writing files


```python
with open('data.txt','w') as file:
    file.write("Hello world")
```

```python
try:
    with open('data.txt','r') as file:
        print(file.read())
except FileNotFoundError:
    print("Invalid File Path")
```

PART-1: READING + WRITING Files(Combined)

Most real application read data, process it and write output.


Example Use Cases:
* Read Student marks --> Write Result
* Read Logs ---> write report
* Read input file --> Generate output files



Example: Read from one file and write input.text

```python
with open("infile.txt","r") as infile:
    content = infile.read()
    # print(content)

with open("output.txt",'w') as outfile:
    outfile.write("Rhymes: \n")
    outfile.write(content)
    print("Data Writes successfully!!")
```

### PART - 2 CSV File Writing in Python

What is CSV?

* Comma Seperated Value
* Used in Excel,database, reports

#### Required Library

import csv

```python
import csv
```

```python
with open("student.csv","w") as file:
    writer = csv.writer(file)
    writer.writerow(["ID","Name","Marks"])
    writer.writerow([1,"Pratiksha",99])
    writer.writerow([2,"Diksha",65])
    writer.writerow([3,"Disha",55])
```

```python
wb = Workbook()

sheet = wb.active

### create column in excel file
sheet["A1"] = "ID"
sheet["B1"] = "Name"
sheet["C1"] = "Salary"


### Add data in the column filed

sheet.append([1,"Ram",900000])
sheet.append([2,"Ramesh",800000])
sheet.append([3,"Shyam",3800000])

#### save data with file name
wb.save("employee.xlsx")
```
