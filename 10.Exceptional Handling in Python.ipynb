## 10. Exceptional Handling in Python

Exceptional handling in Python is used to handle `runtimeError` so that the program does not crash suddenly and can contineu execution smoothly.

In real-world software development, excptional handling is very important becuase errors can occurs anytime due to:

* Wrong user input
* missing file
* database failure
* network issue
* division by zero
* Invalid operations

4. Exceptional Handling Keywords

   Python Provides these keywords in Exceptional Handling.



|Keyword|Meaning|
|----|----|
|`try`|Code that may generate exception|
|`except`| handling exception if any occurs|
|`else`| runs if no errors occur|
|`finally`|always runs|
|`raise`| manually generate exception|
|`assert`| testing conditions raise AssertionError|

### 3. Multiple Except Blocks

You can handle multiple errors at a time.

````python
try:
    fruits = ["apple","banana","mango"]
    print(fruits)
    userInput=int(input("Enter any value: "))
    print(fruits[userInput])
except IndexError:
    print("IndexError")
except ValueError:
    print("ValuError")
except TypeError:
    print("TypeError")
````

## 8. Single Except Hanlidng Multiple Exception



````python
try:
    fruits = ["apple","banana","mango"]
    print(fruits)
    userInput=int(input("Enter any value: "))
    print(fruits[userInput])
except IndexError,ValueError,TypeError:
    print("IndexError/ValueError/TypeError")
````

### 9. Using Exception as Object(as e)

This is used to display actual message on the screen generated by cpu while execution

````python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    val = num1/num2
    print(f"Result:{val}")
except Exception as e:
    print(f"Error:{e}")
````

### 10. try-except-else

````python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    val = num1/num2 
except Exception as e:
    print(f"Error:{e}")

else:
    print(f"Result:{val}")
````

### 11. try-except-else-finally

````python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    val = num1/num2 
except Exception as e:
    print(f"Error:{e}")

else:
    print(f"Result:{val}")

finally:
    print("Execution completed")
````
