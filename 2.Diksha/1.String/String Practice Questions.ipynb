### 4.Count the number of vowels in a string.

```python
name = input()
count = 0

for i in name:
    if i in ("aeiouAEIOU"): #membership operator
        count+=1
        
print(f"Number of vowels in {name}: ",count)
```

### 5.Count the number of consonants in a string.

```python
name = input()
count = 0

for i in name:
    if i not in "aeiouAEIOU" and i.isalpha():
        count+=1

print(f"Number of consonants in {name}: ",count)
```

### 9.Count how many times a particular character occurs in a string.

```python
str = input("Enter a String : ")
ch = input("Enter character : ")

count = 0
for i in str:
    if i==ch:
        count+=1
print(f"'{ch}' occurs ",count," times")
```
