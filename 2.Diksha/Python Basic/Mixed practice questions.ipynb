### 4. Write a program to convert a string to lowercase.

```python
name = "DIKSHA"
print(name.lower())
```

### 5. Write a program to count the number of vowels in a string.

```python
name = "Diksha Ingawale"
count = 0

for i in name:
    if i in "aeiouAEIOU":
        count+=1

print(f"Number of vowels in {name} is: ",count)
```

### 6. Write a program to reverse a string.

```python
name = "diksha"
print(name[::-1])
```

### 7. Write a program to check whether a given string is a palindrome.

```python
s1 = "nayan"
s2 = s1[::-1]

if s1 == s2:
    print("Given string is palindrome")
else:
    print("Given string is not palindrome")
```

### 8. Write a program to count the occurrence of a specific character in a string.

```python
string = "abbdbcbbbsbsbbs"
count = 0

for i in string:
    if i == "b":
        count+=1

print(f"b occurs ",count," times")
```

### 9. Write a program to replace a word in a string with another word.

```python
name = "Hello World"
print(name.replace("World","Python"))
```

### 10. Write a program to split a sentence into a list of words.

```python
sentence = """
Hello
my name is Diksha
"""

print(sentence.split())
```
