```python
## 33. Find the GCD/HCF of two numbers using a while loop.
num1 = int(input("enter 1st number: "))
num2 = int(input("enter 2nd number: "))

gcd = 0
i = 1 
while i <= (min(num1,num2)):
    if num1%i == 0 and num2%i == 0:
        gcd = i
    i+=1
print(f"GCD of {num1} and {num2} is {gcd}")
    
```

```python
### 45. Create a menu-driven calculator using while:

# Addition
# Subtraction
# Multiplication
# Division
# Exit

num1 = int(input("enter no.1 : "))
num2 = int(input("enter no.2 : "))

while True:
    print("1. Addition")
    print("2. Substraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = int(input("enter your choice: "))
    match(choice):
        case 1:
            add = num1 + num2
            print(f"Addition of {num1} and {num2} is {add}")
        case 2:
            sub = num1 - num2
            print(f"Substraction of {num1} and {num2} is {sub}")
        case 3:
            mul = num1 * num2
            print(f"Multiplication of {num1} and {num2} is {mul}")
        case 4:
            div = num1 / num2
            print(f"Division of {num1} and {num2} is {div}")
        case 5:
            break
        case _:
            print(f"Invalid choice!")
```

```python
# 47. Create a program that accepts numbers continuously and displays:
# Total numbers
# Sum
# Average
# Largest number
# Smallest number Stop when the user enters 0.
    
num = int(input("enter number:"))

total = 0
count = 0
maxval = num
minval = num

while num != 0:
    total+=num
    count+=1

    if num >= maxval:
        maxval = num

    if num < minval:
        minval = num

    num = int(input("enter number: "))

print("----------------------")
print("Total numbers : ",count)
print("Total sum : ",total)
print("Largest number : ",maxval)
print("Smallest number : ",minval)
    
```

```python
#50. Check whether a number is an Armstrong number using a while loop for digit processing.***
num = int(input("enter number: "))

temp = num
original = num
digits = 0

while temp > 0:
    digits += 1
    temp = temp//10

temp = num
sum1 = 0
while temp > 0:
    digit = temp % 10
    sum1 = sum1 + digit**digits
    temp = temp//10

if sum1 == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
    
```
