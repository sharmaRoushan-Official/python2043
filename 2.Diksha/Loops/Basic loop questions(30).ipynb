### 1. Print numbers from 1 to 10 using a for loop.

```python
#1. Print numbers from 1 to 10 using a for loop.

for i in range(1,11):
    print(i)
```

### 3. Print the table of a given number (e.g., input 5 → 5, 10, 15, … 50).


```python
#3. Print the table of a given number (e.g., input 5 → 5, 10, 15, … 50).

num = int(input("Enter number: "))

for i in range(1,11):
    print(f"{num} x {i} = {i*num}")
```

### 4. Calculate the sum of numbers from 1 to n.


```python
#4. Calculate the sum of numbers from 1 to n.

n = int(input("Enter number: "))
total = 0

for i in range(1,n+1):
    total+=i
    
print(total)
```

### 5. Calculate the factorial of a number using a loop.


```python
#5. Calculate the factorial of a number using a loop.
n = int(input("Enter number: "))
fact = 1

for i in range(1,n+1):
    fact = fact * i
    
print(fact)
```

### 7. Print the reverse of a given number (e.g., 123 → 321).


```python
#7. Print the reverse of a given number (e.g., 123 → 321).
num = int(input("Enter number:"))
reverse = 0

for i in range(len(str(num))):
    digit = num % 10       #take last digit
    reverse = reverse * 10 + digit   #store
    num = num // 10        #remove last digit
    
print(reverse)
```

### 8. Count digits in a number using a loop.


```python
#8. Count digits in a number using a loop.
num = int(input("Enter number:"))
count =0

for i in range(num):
    if num > 0:
        num = num//10
        count = count + 1
    else:
        break

print("Number of digits = ",count)
    
```

### 14. Find the sum of digits of a number using a while loop.


```python
num = int(input("enter number: "))
temp = num
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print(f"sum of digits of {temp} is {sum}")
```

### 15. Reverse a string using a while loop.


```python
text = input("Enter string: ")
reverse = ""

i = len(text)-1
while i >= 0:
    reverse = reverse + text[i]
    i = i-1

print(reverse)
```

### 27. Check if a string is a palindrome using a loop.


```python
name = input("enter text: ")
reverse = ""

for i in name:
    reverse = i + reverse

if name == reverse:
    print("Palindrome")
else:
    print("not palindrome")
```

### 28. Print Fibonacci series up to n terms using a loop.


```python
n = int(input("Enter number: "))
num1 = 0
num2 = 1
print(num1,num2, end = " ")

for i in range(2,n):
    total = num1 + num2

    num1 = num2
    num2 = total
    print(total,end=" ")
```

### 29. Find the GCD of two numbers using a loop.

```python
num1 = int(input("enter num1: "))
num2 = int(input("enter num2: "))

gcd = 1

for i in range(1,min(num1,num2)+1):              #because the GCD can never be greater than the smaller of the two numbers.
    if num1%i == 0 and num2%i == 0:
        gcd = i

print(gcd)
```
