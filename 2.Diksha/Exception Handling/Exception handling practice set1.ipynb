### **2. Multiple Except Blocks**

Write code that divides two numbers entered by the user. Handle:

* `ValueError` (invalid input)
* `ZeroDivisionError` (denominator = 0)


````python
try:
    a = int(input("enter number1: "))
    b = int(input("enter number2: "))

    div = a/b
    print(div)
    
except ZeroDivisionError:
    print("Error occurred: division by 0")
    
except ValueError:
    print("Please enter valid input")
````

### **3. Using Else Block**

Write a program where:

* `try` block attempts to open a file `data.txt`
* `except` handles `FileNotFoundError`
* `else` prints “File opened successfully”


````python
try:
    file = open("data.txt","r")
except FileNotFoundError:
    print("File not found")
else:
    print("File opened successfully")
    file.close()
````

````python
try:
    file = open("data1.txt","r")
    print(file.read())
except FileNotFoundError:
    print("File not found")
finally:
    print("Operation completed")
````

### **8. Try–Except–Finally With File Closing**

Open a file, read its contents, and ensure it closes in `finally` block even if an error occurs.


````python
file = None
try:
    file = open("poem.txt","r")
    content = file.read()
except FileNotFoundError as e:
    print("File not found: ",e)
else:
    print(content)
finally:
    if file is not None:
        file.close()
        
    print("File closed succesfully")
    
````

### **10. Custom Exception**

Create a custom exception `InvalidMarkError` and raise it when marks > 100 or < 0. Handle it using try-except.


````python
class InvalidMarkError(Exception):
    pass

try:
    marks = int(input("enter marks: "))

    if marks>100 or marks<0:
        raise InvalidMarkError("Marks should be greater than 0 and less than 100")

    print("Marks : ",marks)

except InvalidMarkError as e:
    print("Error:",e)
````

### **15. Try-Except Inside a Function**

Create a function that calculates the square root of a number. Handle ValueError if input < 0.


````python
import math

def sqrnum(num):
    try:
        return math.sqrt(num)
    except ValueError:
        print("Cannot calculate square root of negative number")
    
num = int(input("enter number"))
result = sqrnum(num)

if result is not None:
    print("Square root: ",result)
````

### **16. Reading From a Possibly Corrupt File**

Write a program that tries to read and convert each line of a file into an integer. Handle:

* FileNotFoundError
* ValueError (line cannot convert)


````python
file = open("data.txt", "w")
file.write(str(10))
file.write("\n"+str(100))
file.write("\nabc")

file.close()

try:
    file = open("data.txt","r")
    try:
        for line in file:
            num = int(line.strip())
            print(num)
    except ValueError:
        print("value error: ",line.strip())
    file.close()
    
except FileNotFoundError:
    print("file not found")
````

### **17. Finally for Cleanup**

Simulate allocating a resource (print “Resource allocated”) and always free it in `finally` (“Resource released”) even if an exception occurs.


````python
try:
    print("Resource allocated")
    raise ValueError("Something went wrong")
    
except ValueError as e:
    print("Error : ",e)
    
finally:
    print("Resource released")
````
