### 15. Menu-Driven Calculator

Create a menu-driven calculator using exception handling.

Handle:

- `ValueError`
- `ZeroDivisionError`
- Invalid menu choices

---

```python
while True:
    print("Calculator")
    print("1. Addition")
    print("2. Substraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Reminder")
    print("6. Exit")
    
    try:
        choice = int(input("enter your choice: "))
        
        num1 = int(input("enter 1st number: "))
        num2 = int(input("enter 2nd number: "))

        match(choice):
            
            case 1:
                add = num1+num2
                print(add)
            case 2:
                sub = num1-num2
                print(sub)
            case 3:
                mul = num1*num2
                print(mul)
            case 4:
                div = num1/num2
                print(div)
            case 5:
                mod = num1%num2
                print(mod)
            case 6:
                break
            case _:
                print("Invalid choice")
                continue
    except ValueError:
        print("Invalid input")
    
    except ZeroDivisionError:
        print("cannot divide by zero")
```

### 24. Bank Withdrawal

Create a bank withdrawal program that raises an exception when the withdrawal amount is greater than the available balance.

---

```python
class InsufficientBankBalance(Exception):
    pass

avail_balance = 5000
try:
    withdrawAmount = int(input("Enter withdraw Amount : "))

    if withdrawAmount > avail_balance:
        raise InsufficientBankBalance("withdrawal amount is greater than the available balance.")

    print("Withdrawal Successful")
    print("Remaining balance : ",avail_balance - withdrawAmount)
    
except InsufficientBankBalance as e:
    print(e)
```

### 27. Square Root Validation

Create a function to calculate the square root of a number and raise an exception if the number is negative.

---

```python
import math

class NegativeNumberException(Exception):
    pass

def square_root(num):
    if num < 0:
        raise NegativeNumberException("cannot calculate square root of negative number")
    return math.sqrt(num)

try:
    num = int(input("enter number: "))
    result = square_root(num)
    print(result)

except NegativeNumberException as e:
    print(e)

except ValueError as e:
    print(e)
```

### 29. Student Result Program

Create a student result program using custom exceptions for:

- Invalid marks
- Missing subjects

---

```python
class InvalidMarksError(Exception):
    pass

class MissingSubjectError(Exception):
    pass

def student_result(marks):
    required_subject = ["python","sql","maths"]

    for subject in required_subject:
        if subject not in marks:
            raise MissingSubjectError(f"{subject} marks is missing")


    for subject,mark in marks.items():
        if mark <0  and mark > 100:
            raise InvalidMarksError(f"Invalid marks for {subject}:{marks}")
            
    total = sum(marks.values())
    percentage = total / len(marks)

    print("Total : ",total)
    print("Percentage : ",percentage)

try:
    marks = {
        "python" : int(input("enter python marks: ")),
        "sql" : int(input("enter SQL marks: ")),
        "maths" : int(input("enter math marks: "))
    }
    student_result(marks)
    
except InvalidMarksError as e:
    print(e)

except MissingSubjectError as e:
    print(e)

```

### 30. Mini ATM Program

Build a mini ATM program using:

- `try`
- `except`
- `else`
- `finally`
- `raise`
- Custom exceptions

The program should handle situations such as:

- Invalid PIN
- Insufficient balance
- Invalid withdrawal amount
- Invalid deposit amount
- Invalid user input

---


```python
class InvalidPINError(Exception):
    pass

class InsufficientBalanceError(Exception):
    pass

class InvalidWithdrawAmountError(Exception):
    pass

class InvalidDepositAmountError(Exception):
    pass

class InvalidInputError(Exception):
    pass


def mini_atm():
    balance = 5000
    correct_pin = "1234"
    attempts = 3

    print("======WELCOME TO MINI ATM======")
    
    while attempts > 0:
        try:
            pin = input("enter pin number: ")
            if not pin.isdigit() or len(pin)!=4:
                raise ValueError("PIN must contain exactly 4 digits.")
                
            if pin != correct_pin:
                raise InvalidPINError("Invalid PIN! please enter correct Pin number.")
    
            print("PIN verified successfully")
            break 
            
        except InvalidPINError as e:
            attempts -= 1
            print(e)
            print("Attempts remaining: ",attempts)
    
        except ValueError as e:
            attempts -= 1
            print("Input Error: ",e)
            print("Attempts remaining: ",attempts)

        else:
            print("PIN verified successfully")
            break
            
    else:
        print("Too many incorrect attemps. Card blocked!")
        return

    while True:
        print("------------------")
        print("-----ATM MENU-----")
        print("------------------")
        
        print("1. Check Balance ")
        print("2. Withdraw ")
        print("3. Deposit ")
        print("4. Exit ")
        
        try:
            choice = int(input("enter your choice: "))
            match(choice):
                case 1: 
                    if balance > 0:
                        print("Current balance: ",balance)
                    else:
                        raise InsufficientBalanceError("Balance should be greater than 0")   
        
                case 2:
                    amount = int(input("enter amount to withdraw"))
        
                    if amount <= balance:   
                        balance -= amount
                        print("Balance after withdrawal: ",balance)
                    else:
                        raise InvalidWithdrawAmountError("Amount should be less than balance")
                        
                case 3:
                    amount = int(input("enter amount to withdraw"))
                    
                    if amount > 0:   
                        balance += amount
                        print("Balance after deposit: ",balance)
                    else:
                        raise InvalidDepositAmountError("Amount should be greater than 0")            
                    
                case 4:
                    print("Exit...")
                    break

                case _:
                    print("Invalid Choice! please enter valid choice")
    
        except InvalidPINError as e:
            print(e)
        
        
        except InsufficientBalanceError as e:
            print(e)
        
        
        except InvalidWithdrawAmountError as e:
            print(e)
        
        
        except InvalidDepositAmountError as e:
            print(e)
        
        
        except InvalidInputError as e:
            print(e)

        finally:
            print("ATM transaction process completed")

mini_atm()
```
