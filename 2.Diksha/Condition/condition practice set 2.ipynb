#### 3.A company will decided to give bonus to the employee according to the following criteria:
```
    Time period of the service -------------------> Bonus
    More than 10 Years ---------------------------> 10%
    >=6 and <=10 ---------------------------------> 8%
    <6 -------------------------------------------> 5% 
```
Ask the user for their salary and yers of service and then calcualte the bonus amount is, and display the total salary after increment of bonus.

````python
service_years = int(input("Enter time period of service: "))
salary = float(input("Enter your salary: "))

if service_years > 10:
    bonus = salary * 10 / 100
    TotalSalary = salary + bonus

elif service_years >= 6 and service_years <= 10:
    bonus = salary * 8 / 100
    TotalSalary = salary + bonus

else:
    bonus = salary * 5 / 100
    TotalSalary = salary + bonus

print("Bonus = ",bonus)
print("Total salary = ",TotalSalary)
````

#### 7. Accept the following from the user and calculate the percentage of the class attendence.
```
Total number of working days.
Total number of absent.
if percentage is less than 75%, then student will not sit the exam.

````python
WorkingDays = int(input("enter no. of working days: "))
absentDays = int(input("enter number of absent students: "))

percentDays = WorkingDays - absentDays

percentage = (percentDays / WorkingDays) * 100

print("Attendance percentage : ",percentage)
if percentage < 75 :
    print("Student will not sit for the exam")
else:
    print("Student will sit for the exam")
    
````

#### 15. Crete a simple calulator that performs basic arithmetic operations based on two user input.(^,*,/,-,+)

````python
n1 = int(input("enter num1: "))
n2 = int(input("enter num2: "))

while True:
    print("1. Addition: ")
    print("2. Substraction: ")
    print("3. Multiplication: ")
    print("4. Division: ")
    print("5. Power: ")
    print("6. Exit: ")

    choice = int(input("enter your choice: "))
    
    match(choice):
        case 1 :
            add = n1+n2
            print(f"Addition of {n1} and {n2} is {add}")

        case 2 :
            sub = n1-n2
            print(f"Substraction of {n1} and {n2} is {sub}")

        case 3 :
            mul = n1*n2
            print(f"Multiplication of {n1} and {n2} is {mul}")

        case 4 :
            div = n1/n2
            print(f"Division of {n1} and {n2} is {div}")

        case 5 :
            pow = n1**n2
            print(f"Power of {n1} and {n2} is {pow}")

        case 6 :
            break

        case _:
            print("Invalid choice!")

            
````
