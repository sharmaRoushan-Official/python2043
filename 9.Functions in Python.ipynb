````python
def sumTwo(a,b=0):  # b= 0 is default arguments
    return f"The sum of two numbers {a} and {b} is {a+b}"

sumTwo(-14,25)
````

## Varaible length argument 
1. *args
2. **kwargs

````python
def sumNum(*num):
    return f"Sum of the numbers {num} is:{sum(num)}"
sumNum(10)
````

````python
sum([10,20])
````

````python
10,
````

````python
def sumNum(*num):
    return f"Sum of the numbers {num} is:{sum(num)}"
print(sumNum(10))
print(sumNum(10,10))
print(sumNum(10,10,45))
print(sumNum(10,10,45,78))
print(sumNum(10,10,45,78,45))
````

````python
def sumNum(*args):
    return f"Sum of the numbers {args} is:{sum(args)}"
print(sumNum(10))
print(sumNum(10,10))
print(sumNum(10,10,45))
print(sumNum(10,10,45,78))
print(sumNum(10,10,45,78,45))
````

## **kwargs --> keyword arguments

````python
def personDetails(**kwargs):
    return f"Details of person is: {kwargs}"

print(personDetails(name="Pratiksha"))
print(personDetails(name="Pratiksha",age=25))
print(personDetails(name="Pratiksha",age=25,course="python"))
````

````python
# 2. local 
def test1():
    x1 = 100 # local varaible/scop : it's scope is only accessible inside the functions
    return x1
test1()
````

### 9. Nested Functions 

A function inside another functions


````python
def outerFunc():
    print("Outer Function is calling")
    def innerFunc():
        print("Inner Function is calling")
        def inner2Func():
            print("Inner 2 Function is calling")
        inner2Func()
    innerFunc()
outerFunc()
````

### 11. Docstring
