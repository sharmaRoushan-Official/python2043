### 18. Define an abstract class FileHandler with methods read() and write(). Create child classes TextFileHandler and BinaryFileHandler.

```python
from abc import ABC,abstractmethod

class FileHandler(ABC):

    @abstractmethod
    def read(self):
        print("read the file")

    @abstractmethod
    def write(self):
        print("write the file")

class TextFileHandler(FileHandler):
    def read(self):
        print("read the file")

    def write(self):
        print("write the file")

    def text(self):
        print("Text the file")

class BinaryFileHandler(FileHandler):
    def write(self):
        print("write the file")

    def read(self):
        print("read the file")

    def binary(self):
        print("file is binary")

obj = TextFileHandler()
obj.read()
obj.write()
obj.text()

obj1 = BinaryFileHandler()
obj1.read()
obj1.write()
obj1.binary()
    
```
