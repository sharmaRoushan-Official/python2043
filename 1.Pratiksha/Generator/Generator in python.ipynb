## How Generator works

Steps:

1. Function is called -> No Execution
2. Return generator obj
3. next() resumes execution
4. Stop at yield
5. continues from same point


## Generator VS Normal Function

|Feature|Generator|Function|
|----|---|----|
|Used|`yield`|`return`|
|Memory|`low`|`High`|
|Execution|`Lazy`|`Immediate`|
|Values|`One at a time`|`All at once`|

## Reading files using generator

```python
def readfile(fileName):
    with open(fileName,'r') as file:
        for line in file:
            yield line.strip()
             
```

### When to use Generator

* Large Dataset
* infinite sequence
* streaming data
* Performance critical apps

### When not to use Generator

* Need Random access
* Need all value repeatedly

* Generator uses yield
* Memory efficient
* Produces value lazily
* next() fetches the next value

#### 6️⃣ Write a generator that yields the Fibonacci sequence up to N terms.

```python
n=7
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for i in fibonacci(n):
    print(i, end=" ")
        
```

#### 7️⃣ Write a generator that yields the factorials of numbers from 1 to N.

```python
import math
def factorial(n):
    for i in range(1,n):
        yield math.factorial(i)

for i in factorial(10):
    print(i, end=" ")
```

#### 8️⃣ Write a generator that yields all prime numbers between 1 and N.

```python
def prime(n):
    for i in range(2, n+1):
        for j in range(2, i):
            if i % j == 0:
                break
        else:
            yield i

for i in prime(20):
    print(i, end=" ")    
```

#### 9️⃣ Write a generator that yields each character of a given string one by one.

```python
def characters(s):
    for ch in s:
        yield ch

for i in characters("Python"):
    print(i)
```

#### 🔟 Write a generator that yields each word from a given sentence.

```python
def sentence(s):
    for word in s.split():
        yield word

for i in sentence("I am Pratiksha"):
    print(i)
```

#### 1️⃣4️⃣ Write a generator that yields the running sum of numbers in a list.
Example:
Input: [1, 2, 3, 4]
Output: 1, 3, 6, 10

```python
def runningSum(nums):
    total = 0
    for i in nums:
        total = total + i
        yield total

for i in runningSum([1, 2, 3, 4]):
    print(i)
```

#### 1️⃣6️⃣ Create an infinite generator that keeps yielding natural numbers starting from 1.

```python
def natural():
    i = 1
    while True:
        yield i
        i += 1

g = natural()

for i in range(10):
    print(next(g))
```

#### 1️⃣7️⃣ Write a generator that yields all palindrome numbers between 1 and N.

```python
def palindrome(n):
    for i in range(1, n + 1):
        if str(i) == str(i)[::-1]:
            yield i

for i in palindrome(20):
    print(i)
```


#### 1️⃣8️⃣ Write a generator that yields the binary representation of numbers from 1 to N.

```python
def binary(n):
    for i in range(1, n + 1):
        yield bin(i)[2:]

for i in binary(5):
    print(i)
```

#### 1️⃣9️⃣ Write a generator that yields numbers from 1 to N:
✔️ Divisible by 3 → "Fizz"

✔️ Divisible by 5 → "Buzz"

✔️ Divisible by both → "FizzBuzz"

```python
def fizzbuzz(n):
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            yield "FizzBuzz"
        elif i % 3 == 0:
            yield "Fizz"
        elif i % 5 == 0:
            yield "Buzz"
        else:
            yield i

for i in fizzbuzz(15):
    print(i)
```

#### 2️⃣0️⃣ Write a generator that accepts a list and yields only unique elements while preserving the original order.
Example:
Input: [1, 2, 2, 3, 4, 4, 5]
Output: 1, 2, 3, 4, 5

```python
def unique(nums):
    seen = set()
    for i in nums:
        if i not in seen:
            seen.add(i)
            yield i

for i in unique([1, 2, 2, 3, 4, 4, 5]):
    print(i)
```
