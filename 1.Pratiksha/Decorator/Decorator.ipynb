```python
def decorator(func):
    def wrapper():
        print("Before Function")
        func()
        print("After Function")
    return wrapper
```

### Decorator with *args and **kwargs

```python
def decorator(func):
    def wrapper(*args,**kwargs):
        print("function started")
        result = func(*args,**kwargs)
        print("function ended")
    return wrapper
```

```python
import time
def decorator(func):
    def wrapper(*args,**kwargs):
        print("function started")
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print("Execution Time:",end-start)
        print("function ended")
    return wrapper
```

```python
@decorator
def demo():
    for i in range(1000000000):
        pass

demo()
```

## 4. Function Call Counter
Create a decorator that counts and displays how many times a function has been called.

```python
def decor(func):
    count = 0

    def wrapper():
        nonlocal count
        count += 1
        print("Function called", count, "times")
        func()

    return wrapper


@decor
def hello():
    print("Hello World")


hello()
hello()
hello()
hello()      
```


## 8. Even Number Decorator
Create a decorator that checks whether the given number is even before executing the function. If the number is odd, print "Only even numbers are allowed."

```python
def dec(func):
    def wrapper(num):
        if num % 2 == 0:
            return func(num)
        else:
            print("Only even numbers are allowed.")
    return wrapper


@dec
def even_odd(num):
    print("Given number is even")


even_odd(20)
even_odd(15)
```


## 9. Password Protection Decorator
Create a decorator that asks the user for a password before executing the function. If the password is correct, execute the function; otherwise print "Access Denied."

```python
def decorate(func):
    def wrapper():
        password = "1234"
        user_password = input("Enter the password: ")
        if password == user_password:
            return func()

        else:
            print("Access Denied")
    return wrapper

@decorate
def password():
    print("Password is correct")
password()
        
```

## 10. Exception Handling Decorator
Create a decorator that catches any exception raised by a function and prints the exception message instead of stopping the program.

```python
def exception_handler(func):
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except Exception as e:
            print("Exception occurred:", e)

    return wrapper


@exception_handler
def divide(a, b):
    return a / b


print(divide(10, 2))
divide(10, 0)

print("Program continues...")
```


## 16. Input Validation Decorator
Create a decorator that checks if all arguments passed to a function are integers. If not, print "Invalid Input."

```python
def validate_input(func):
    def wrapper(*args):
        for i in args:
            if type(i) != int:
                print("Invalid Input")
                return
        return func(*args)
    return wrapper

@validate_input
def add(a, b):
    print("Addition:", a + b)

add(10, 20)
add(10, "hello")
add(5, 6.5)
        
```


## 19. Debug Decorator
Create a decorator that prints the function name, arguments, and returned value whenever the function is called.

```python
def debug(func):
    def wrapper(*args):
        print("Function Name:", func.__name__)
        print("Arguments:", args)

        result = func(*args)

        print("Returned Value:", result)
        return result

    return wrapper


@debug
def add(a, b):
    return a + b


print(add(10, 20))
```


## 20. Retry Decorator
Create a decorator that retries a function up to 3 times if it raises an exception before finally displaying the error.

```python
def retry(func):
    def wrapper():
        for i in range(3):
            try:
                return func()
            except Exception as e:
                print("Attempt", i + 1, "failed")

        print("Error:", e)

    return wrapper


@retry
def divide():
    return 10 / 0


divide()
```
