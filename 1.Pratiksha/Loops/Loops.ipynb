## While loop

Syntax: 

while condition:
  true block

```````python
# 0-10

i = 0
j = 10

while i<j:
    print(i)
    i = i+1
```````

```````python
#### Decreasing order

i = 10
j = 0

while i > j:
    print(i)
    i = i-1
```````

```````python
## table of 18

i = 1
j = 10
num = 18
while i <= j:
    print(num,"*",i,"=",num*i)
    i+=1
```````

```````python
# table of 5 to 10

i = 5
j = 10

while i<=j:
    # print(i)
    a = 1
    b = 10
    while a <= b:
        print(i,"*",a,"=",i*a)
        a+=1
    print()
    i+=1
```````

```````python
i = 1
j =10
sum1 = 0


while i <= j:
    sum1+=i
    i+=1
    
print("Sum of the values from 1-20:",sum1)
```````

```````python
i = 1
j = 20

while i <= 20:
    if i % 2 == 0:
        print(i)
    i+=1
```````

```````python
i = 1
j = 10
sum1 = 0

while i <= 10:
    if i % 2 == 0:
        sum1+=i    
    i+=1
print("Sum of even numbers from 1-10:",sum1)
```````
