```python
sum1 = 0
list1 = []
while True:
    num = int(input("Enter the number:"))
    if num == 0:
        break
    list1.append(num)
    sum1+=num
print("User's Numbers is:",list1)
print("Sum of the user's value is:",sum1)
```

# 22.Take a number from the user and find the sum of its digits.

```python
number = int(input("Enter the number:"))
sum1 = 0

while number != 0:
    digit = number % 10
    number = number // 10
    sum1 += digit
print("Sum of digits:",sum1)
```

# 23. Take a number from the user and find the product of its digits.

```python
number = int(input("Enter the number:"))
product = 1

while number != 0:
    digit = number % 10
    number = number // 10
    product *= digit
print("Product of digits:",product)
```

# 24.Reverse a number using a while loop.

```python
number = input("Enter the number: ")
i = len(number) - 1
revNumber = ""
while i >= 0:
    revNumber += number[i]
    i -= 1
print("Reverse Number:", revNumber)
```

# 25. Check whether a number is a palindrome.



```python
number = input("Enter the number: ")

i = len(number) - 1
revNumber = ""

while i >= 0:
    revNumber += number[i]
    i -= 1

if number == revNumber:
    print("The number is palindrome")
else:
    print("The number is not palindrome")
```
