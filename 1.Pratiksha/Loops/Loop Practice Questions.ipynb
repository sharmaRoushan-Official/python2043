# For loop Questions

# 1.Print numbers from 1 to 10 using a for loop.

```python
for i in range(1,11):
    print(i)
```

# 2. Print all even numbers between 1 and 100

```python
for x in range(0,111,2):
    print(x,end=" ")
```

# 3. Print the table of a given number (e.g., input 5 → 5, 10, 15, … 50).

```python
num =5

for i in range(1,11):
    print(num,"*",i,"=",num*i)
```

# 4. Calculate the sum of numbers from 1 to n

```python
sum = 0

for i in range(100):
    sum = sum + i
print(sum)
```

# 5.Calculate the factorial of a number using a loop

```python
num = 5
value = 1
for i in range(1,num+1):
    value = value * i
print(value)
```

# 6. Print all numbers divisible by 3 and 5 from 1 to 100

```python
for i in range(1,101):
    if i % 3 == 0 and i % 5 == 0:
        print(i)
```

# 7. Print the reverse of a given number (e.g., 123 → 321)

```python
number = "123"

revNum = ""
for val in number:
    revNum = val + revNum
print(revNum)
    
```

# 8. Count digits in a number using a loop

```python
number = "1234"
count = "0"
for i in number:
    count = count + i
print(count)
```

# 9. Print all characters of a string using a loop

```python
character = "Pratiksha"
string = ""
for i in character:
    string = string + i
print(string)
```

# 10. Print only vowels from a string input by the user

```python
string = input("Enter the string:")
vowel = "aeiouAEIOU"

for i in string:
    # print(i)
    if i in vowel:
        print(i)

```
