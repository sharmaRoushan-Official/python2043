# List Comprehension Practice questions
## 1.Generate a list of squares from 1 to 10 using list comprehension.

```python
[i**2 for i in range(1,11)]
```


## 2. Create a list of even numbers from 1 to 50 using list comprehension.

```python
even = [i for i in range(1, 51) if i % 2 == 0]

print(even)
```


## 3. Convert the list of strings to uppercase using list comprehension:
    words = ['cat', 'dog', 'lion']


```python
words = ['cat', 'dog', 'lion']

result = [word.upper() for word in words]

print(result)
```


## 4. Get a list of numbers divisible by both 3 and 5 from 1 to 100.

```python
list = [i for i in range(1,101) if i % 3 == 0 & i % 5 == 0]
print(list)
```


## 5. Extract all vowels from the string using list comprehension:
    text = "Comprehension"

```python
text = "Comprehension"

vowels = [ch for ch in text if ch in "aeiouAEIOU"]

print(vowels)
```


## 6. Given a matrix:
    matrix = [[1,2,3],[4,5,6],[7,8,9]]
    Use list comprehension to flatten it into a single list.

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flatten_list = [j for i in matrix for j in i]
flatten_list
```


## 7. Given a list of words, extract words that contain the letter ‘e’:
    words = ['tree', 'book', 'python', 'sky']

```python
words = ['tree', 'book', 'python', 'sky']

result = [word for word in words if 'e' in word]

print(result)
```


## 8. Create a list of all 3-digit numbers that are palindromes using list comprehension.

```python
palindromes = [i for i in range(100, 1000) if str(i) == str(i)[::-1]]

print(palindromes)
```


## 9. From a list of numbers, replace odd numbers with 'ODD' and keep even numbers as is:
    nums = [1, 2, 3, 4, 5, 6]

```python
nums = [1, 2, 3, 4, 5, 6]

result = ['ODD' if i % 2 != 0 else i for i in nums]

print(result)
```


## 10. Using nested list comprehension, generate a 3x3 identity matrix.

```python
matrix = [[1 if i == j else 0 for j in range(3)] for i in range(3)]

for row in matrix:
    print(row)
```
