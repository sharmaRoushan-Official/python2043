```python
value = input("Enter value: ")

if ord(value) == 32:
    print("You entered Space")
elif ord(value)>=65 and ord(value)<=90:
    print(f"You entered upper case: {value}")
elif ord(value)>=97 and ord(value)<=122:
    print(f"You entered lower case: {value}")
elif ord(value)>=48 and ord(value)<=57:
    print(f"You entered digit value: {value}")
elif ord(value)>=33 and ord(value)<=47 or ord(value)>=58 and ord(value)<=67:
    print(f"You entered special characters: {value}")
```
