# 4.Count the number of vowels in a string.

```python
s="Samdoli"
count =0

for char in s:
    if char in "aeiou":
        count = count+1
print(count)
```

# 5.Count the number of consonants in a string

```python
s="Maharashtra"
count =0

for char in s:
    if char.isalpha() and char not in "aeiou":
        count = count+1
print(count)
    
```
