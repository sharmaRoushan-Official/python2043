### Expression
----
Q. What is an Expression?

----

An `expression` is a combination of:

  * variables
  * values
  * operators
  * function call that produce result.
----

1. Arithematic Expression

Operator:

|Operator|Meaning|
|----|----|
|`+`|Addition|
|`-`|Subtraction|
|`*`|Multiplication|
|`/`|Division|
|`//`|Floor Division|
|`%`|Modulus|
|`**`|Exponent or Power|

```python
# Example: 

a = 10
b = 5.63

result = a+b
print(result)

result2 = a-b
print(result2)

result3 = a/b
print(result3)

result4 = a//b
print(result4)

result5 = a**b
print(result5)

result6 = a%b
print(result6)
```
