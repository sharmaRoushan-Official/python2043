### type() 

```python
# 1. String
print("Hello world")
print(type("Hello world"))
print('Hello world')
print("100")
print("100.23")

# 2. Integer

print(100)
print(type(100))
print(456)
print(789)
print(2116556)


# 3. Float
print(100.0)
print(type(100.0))
print(456.23)
print(78.9)


# 4. boolean 
print(True)
print(type(True))
print(False)
print(type(False))
```

## 1. Introduction to Varaibles

A `variable` is a name used to store data in memeory. [RAM-Random Access Memory]

Or

A variable as a labelled container that hold information.

Example

```python
name = "Pratiksha"
age = 44
course = "Python"
salary = 45000.56

print(name)
print(age)
print(course)
```

Notes:

* `name` stores text
* `age` stores integer
* `course` stores text
* `salary` stores decimal value (float)

### 4. Types of Values in Python

Python automatically identifies data types.


A. Integer(int)

B. Float(float)

C. String(str)

D. Boolean(bool)

```python
# A. Integer

age = 30
print(age, type(age))


# B. Float(float)

price = 99.99
print(price, type(price))


# C. String(str)
city = "NOIDA"
print(city,type(city))


# D. Boolean(bool)
is_active = True
print(is_active,type(is_active))
```

### 5. Dynamic Typed langauge
